## Merging the Data 
This section of my program loads both the Hospital and Personnel Data and inspects the tables, and then merges the two data frames on the "HospitalID" variable. Then, the specified columns are deleted to create the merged data file before exporting. 

```python
# Load Data - Both files and inspect them
# Hospital Data
hospital_data = pd.read_csv("CaliforniaHospitalData.csv")
hospital_data.info()
```

```python
# Personnel Data
personnel_data = pd.read_csv("CaliforniaHospitalData_Personnel.txt", delimiter = "\t")
personnel_data
```

```python
# Merging the data on HospitalID
merged_hospital = hospital_data.merge(personnel_data, how = 'inner', on = "HospitalID")
merged_hospital
```

```python
# Finding duplicate columns- None
merged_hospital.columns
```

```python
# Removing Columns- Work_ID, PositionID, Website
final_hospital = merged_hospital.drop(['Work_ID', 'PositionID', 'Website'], axis = 1)
final_hospital
```

## Exporting the Data
This section of my program involves filtering the data to our interests, which is selecting Small/Rural hospitals that have 15 or more available beds with a positive operating income. This data frame is exported as a tab-delimited file called`hospital_data_new.txt`, which is saved in the directory `data`.

```python
# Final changes to DataFrame: Small/Rural with at least 15 beds and positive operating income
export_data = final_hospital[(final_hospital.Teaching == 'Small/Rural') & (final_hospital.AvlBeds >= 15) & (final_hospital.OperInc > 0) ]
```

```python
# Converting to file in Data folder- See references 
export_data.to_csv('hospital_data_new.txt', sep = "\t", index = False)
```

## Filtering Data
This section of my program filters the data to select all hospitals that are non-profit with more than 250 employees, unless the net patient revenue is smaller than \$109,000. I also remove the columns containing employee information and save it as a new data frame, which is `new_merge2`.
```python
# Hospitals that are non-profit with more than 250 employees, or net patient revenue less than $109,000
new_merge2 = new_merge[((new_merge.TypeControl == 'Non Profit') & (new_merge.FullTimeCount > 250)) | (new_merge.NetPatientRevenue < 109000) ]
new_merge2
```
